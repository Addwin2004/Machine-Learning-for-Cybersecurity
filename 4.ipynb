```python
# Password Strength Classification using Machine Learning
# Algorithm: Decision Tree Classifier

import pandas as pd
import re

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

passwords = [
    # Weak passwords
    "123456",
    "password",
    "12345678",
    "qwerty",
    "abc123",
    "111111",
    "123123",
    "password1",
    "admin",
    "welcome",
    "letmein",
    "iloveyou",
    "football",
    "monkey",
    "dragon",
    "sunshine",
    "master",
    "login",
    "hello123",
    "computer",

    # Medium passwords
    "Password1",
    "Welcome1",
    "Hello@123",
    "Admin123",
    "Qwerty12",
    "Computer1",
    "Football9",
    "India123",
    "Summer2026",
    "Secure123",
    "User@123",
    "Test@2026",
    "MyPassword1",
    "College123",
    "Network@1",
    "Cyber1234",
    "Student@12",
    "Python123",
    "Windows10",
    "Linux@123",

    # Strong passwords
    "P@ssw0rd!X",
    "X7#pL9@qZ2!",
    "M8$kR2!vT9@",
    "Zq7!Lp2#Wx9$",
    "A9@kLm#72!Q",
    "rT8$Yp2@Nq7!",
    "K4!zX9@Lm2#",
    "P7#vR3!xQ8@",
    "N9@wK5$zL2!",
    "Q2!mX8#pR7$",
    "H6@tY9!kP3#",
    "B8#qW2@xM7!",
    "J5!rL9$zT4@",
    "C7@pN3#vX8!",
    "F9$kR4!mQ2@",
    "D6#xP8@wL3!",
    "G3!nZ7$kT9@",
    "S8@vM2#qR6!",
    "W4!pK9$xN7@",
    "Y7#tL3@cQ8!"
]

labels = (
    ["Weak"] * 20 +
    ["Medium"] * 20 +
    ["Strong"] * 20
)

data = pd.DataFrame({
    "password": passwords,
    "strength": labels
})


# STEP 2: Create a function to extract password features

def extract_features(password):

    length = len(password)

    uppercase_count = sum(1 for c in password if c.isupper())

    lowercase_count = sum(1 for c in password if c.islower())

    digit_count = sum(1 for c in password if c.isdigit())

    special_count = sum(
        1 for c in password
        if not c.isalnum()
    )

    # Count repeated characters
    repeated_count = len(password) - len(set(password))

    # Check whether password contains common weak words
    common_words = [
        "password",
        "admin",
        "welcome",
        "qwerty",
        "hello",
        "login",
        "user",
        "test"
    ]

    contains_common_word = int(
        any(word in password.lower() for word in common_words)
    )

    return [
        length,
        uppercase_count,
        lowercase_count,
        digit_count,
        special_count,
        repeated_count,
        contains_common_word
    ]


# STEP 3: Extract features from all passwords

features = data["password"].apply(extract_features)

feature_names = [
    "length",
    "uppercase",
    "lowercase",
    "digits",
    "special_characters",
    "repeated_characters",
    "common_word"
]

X = pd.DataFrame(
    features.tolist(),
    columns=feature_names
)

y = data["strength"]


# Display the dataset
print("\n========== PASSWORD DATASET ==========\n")
print(pd.concat([data, X], axis=1))


# STEP 4: Split data into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)


# STEP 5: Create and train the Decision Tree model

model = DecisionTreeClassifier(
    random_state=42,
    max_depth=5
)

model.fit(X_train, y_train)


# STEP 6: Make predictions on test data

y_pred = model.predict(X_test)


# STEP 7: Calculate accuracy

accuracy = accuracy_score(y_test, y_pred)

print("\n========== MODEL PERFORMANCE ==========")

print("\nAccuracy:", round(accuracy * 100, 2), "%")


# STEP 8: Classification report

print("\nClassification Report:\n")

print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)


# STEP 9: Confusion Matrix

print("Confusion Matrix:\n")

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=["Weak", "Medium", "Strong"]
)

print(cm)


# STEP 10: Predict strength of a new password

print("\n========== PASSWORD STRENGTH PREDICTION ==========")

new_password = input("\nEnter a password to test: ")

new_features = extract_features(new_password)

new_data = pd.DataFrame(
    [new_features],
    columns=feature_names
)

prediction = model.predict(new_data)[0]

print("\nPassword Strength:", prediction)


# STEP 11: Display probability of each class

probabilities = model.predict_proba(new_data)[0]

print("\nPrediction Probabilities:")

for class_name, probability in zip(
    model.classes_,
    probabilities
):
    print(
        class_name,
        ":",
        round(probability * 100, 2),
        "%"
    )


print("\n========================================")
print("Password classification completed.")
print("========================================")
```
